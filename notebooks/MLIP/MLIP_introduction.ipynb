### Message passing neural networks (MPNN)

<div style="font-size: 18px;">

We will briefly introduce the basic terminology and come back to MPNNs when discussing MACE[1].

> Structures are encoded as **graphs** with the atoms as **nodes**. The graph **edges** typically do not represent chemical bonds but are defined within a distance cutoff.

A *fully-connected* graph:

<div style="display: flex; justify-content: left;">
  <img src="../../data/presentation/MLIP_figures/fully_connected.png" style="width: 350px;"/>
</div>
<br>
In the first information exchange (message pass) each node learns about its neighbors within a defined cutoff distance. This leads to the node *features* being updated. Therefore, if another message pass is performed, each node again learns about its neighbors, but all of them have been influenced by their own neighborhood already.

The regions of the graph a node learns about in this way are called its *receptive field*.

### Data splits

> **always** hold out part of your data for testing

We typically split data into three categories: training, validation, and test (often in ratios of `8:1:1` but ymmv). For test data the separation is clear: never use it for anything else to avoid so-called *data leakage*. Basically, if you cannot guarantee that your test data is never used for any training or validation, your results lose meaning and are automatically hard to defend (somewhere, reviewer 2 is furiously taking notes).

A multitude of strategies for splitting training and validation data exist and there is no one-size-fits-all approach. We will not go into that here, just take with you that it can be trivial, but also enable more intricate methods.

> **always** look at your data and workflow before deciding on a strategy

While random splitting is often the correct approach and what we will do in a minute, sometimes it may lead to problems. For example, if your data set is very small, so-called *outliers* can prevent training convergence. As we will see in a bit, MLIPs are typically trained to predict energies and forces. A typical outlier would be a structure with orders of magnitude higher interatomic forces present in the structure.

### The loss function

<div style="font-size: 18px;">

> how do we train the model?

$
\begin{gather}
    \hspace{60px}\mathcal{L} = w_E \cdot \mathcal{L}_E + w_{\boldsymbol{f}} \cdot \mathcal{L}_{\boldsymbol{f}}
\end{gather}
$

A target function (the *loss* $\mathcal{L}$) is minimized over a given number of epochs. The loss is the error (e.g., RMSE) the model makes when predicting a value when compared to the ground truth (often DFT results). For us this typically means multiple loss components that need to be balanced, including energies and forces or other structure properties.

**Training workflow:**
- For each epoch (= loop over all training structures, reshuffled):<br>
    - For each training step (= one batch of `batch_size` structures):<br>
      - Predict energies and forces for all batch structures<br>
      - Calculate the batch loss $\mathcal{L}$ on the training data<br>
      - Update model parameters (optimizer step)<br>
    - Calculate the epoch loss on the validation data<br>
      - Save a checkpoint if the validation loss has improved<br>

In most current MLIP architectures, interatomic forces are not predicted directly, but calculated from the structure energy as $\boldsymbol{f}_i = -\nabla_{\boldsymbol{r_i}}E$. Still, by being able to train on energies (one per structure) and forces ($3\,N$ per $N$-atom structure) greatly improves data availability.

> We decompose the structure into overlapping local environments around each atom.

### Interlude: Invariance vs equivariance

<div style="font-size: 18px;">

Here, we need to briefly discuss the distinction between "invariance" and "equivariance".

> A quantity is **invariant** under a transformation if it does not change. It is **equivariant** if it transforms the same way as the input.

The mentioned descriptors, i.e., the isolated local environments, are invariant under rotation and translation. An emphasis needs to be put on "isolated", e.g., no external field or similar interference.

The total energy of a structure is likewise invariant. Forces on atoms are equivariant: they are vectors and must rotate with the structure. In invariant MLIPs, forces are not predicted directly, they are obtained by differentiating the invariant energy with respect to atomic positions, which is automatically equivariant.

"Early" MLP-based MLIPs achieve invariance by precomputing invariant descriptors before the network ever sees the data. Equivariant architectures such as MACE instead build the symmetry directly into the network, allowing it to operate on vector and tensor quantities explicitly.

Equivariant models retain orientation and track how it transforms. Since rotation symmetry is enforced exactly rather than learned from data, they are significantly more data efficient.

        All model parameters are updated using only the new target dataset. While straightforward, it often leads to "forgetting" the diverse chemical knowledge contained in the original foundation model.
    </td>
  </tr>
  
  <!-- Row 2: Parameter Freezing -->
  <tr>
    <td style="font-size: 1.2em; text-align: left; vertical-align: middle; border: none; padding-bottom: 25px;">
        <b>Parameter freezing</b><br>
        <span style="font-size: 1.0em; color: #666;">low data requirements</span>
    </td>
    <td style="font-size: 1.1em; text-align: left; border: none; padding-bottom: 25px; line-height: 1.5;">
        Locks the model parameters up to a specified layer and trains the rest. This keeps, for example, the learned representation unchanged.
    </td>
  </tr>

  <!-- Row 3: Multihead Tuning -->
  <tr>
    <td style="font-size: 1.2em; text-align: left; vertical-align: middle; border: none; padding-bottom: 25px;">
        <b>Multihead tuning</b><br>
        <span style="font-size: 1.0em; color: #666;">expensive computation</span>
    </td>
    <td style="font-size: 1.1em; text-align: left; border: none; padding-bottom: 25px; line-height: 1.5;">
        Uses a shared backbone to process both new specific data and a "replay" subset of the original foundational data. This preserves broad generalization while achieving high local accuracy.
    </td>
  </tr>

  <!-- Row 4: Delta Learning -->
  <tr>
    <td style="font-size: 1.2em; text-align: left; vertical-align: middle; border: none; padding-bottom: 25px;">
        <b>Delta learning</b><br>
        <span style="font-size: 1.0em; color: #666;">second model or head</span>
    </td>
    <td style="font-size: 1.1em; text-align: left; border: none; padding-bottom: 25px; line-height: 1.5;">
        Instead of learning total energy from scratch, a model learns the residual difference ($\Delta E$) between the pre-trained MLIP and higher-level reference data (e.g., DFT) for a specific system.
    </td>
  </tr>

  <!-- Row 5: Optimization Cycle -->
  <tr>
    <td style="font-size: 1.2em; text-align: left; vertical-align: middle; border: none;">
        <b>All of the above</b><br>
        <span style="font-size: 1.0em; color: #666;">it depends</span>
    </td>
    <td style="font-size: 1.1em; text-align: left; border: none; line-height: 1.5;">
        <b>Single-shot</b> is a one-time update on a fixed dataset. <b>Iterative</b> (active learning) involves multiple cycles of identifying high-uncertainty configurations, labeling them with DFT, and retraining the model.
    </td>
  </tr>
